### What usually goes wrong

These are the five things that sink teams. The charter below is built to prevent each one.

5. **Leaving the weather API to the last week.** It is the one part you can't fake, and Question 3 depends on it.

### Profile the data before you plan anything

You cannot assign work you have not looked at. Run this on the real transactions file. The function is worth keeping — you will run it again on the capstone data in November.

````python
import pandas as pd, os

def find_data(filename, folders=('data', '../data', '/kaggle/input', '/content')):
    for folder in folders:
        path = os.path.join(folder, filename)
        if os.path.exists(path):
            return path
    return None

def profile(df, name='dataset'):
    print(f'=== {name} ===')
    print(f'{len(df):,} rows x {len(df.columns)} columns')
    print()
    summary = pd.DataFrame({
        'dtype': df.dtypes.astype(str),
        'nulls': df.isnull().sum(),
        'null_pct': (100 * df.isnull().mean()).round(1),
        'distinct': df.nunique(),
    })
    print(summary)
    print()
    print('exact duplicate rows:', df.duplicated().sum())
    return summary

path = find_data('walmart_transactions.csv')
if path:
    tx = pd.read_csv(path)
    profile(tx, 'walmart_transactions')
else:
    print('Upload the data/ folder in Kaggle or Colab, or run this inside the repo.')
````

````python
team = {
    'name': 'TODO',
    'members': [                                # one line per person, 2 to 4
        {'name': 'TODO', 'github': 'TODO'},
        {'name': 'TODO', 'github': 'TODO'},
    ],
    'goals': {
        'good_result': 'TODO',                  # what a good result looks like for us
        'what_each_person_wants': 'TODO',       # one line per person is fine
        'time_limits': 'TODO',                  # jobs, other classes, travel
        'what_to_avoid': 'TODO',                # what would make this a bad experience
    },
    'roles': {                                  # the owner makes sure it happens, the backup steps in
        'cleaning + decision log': {'owner': 'TODO', 'backup': 'TODO'},
        'weather API + join':      {'owner': 'TODO', 'backup': 'TODO'},
        'analysis Q1-Q5':          {'owner': 'TODO', 'backup': 'TODO'},
        'charts + presentation':   {'owner': 'TODO', 'backup': 'TODO'},
    },
    'question_owners': {                        # the reviewer is someone other than the owner
        'Q1 demand surge':         {'owner': 'TODO', 'reviewer': 'TODO'},
        'Q2 SKU consolidation':    {'owner': 'TODO', 'reviewer': 'TODO'},
        'Q3 weather correlation':  {'owner': 'TODO', 'reviewer': 'TODO'},
        'Q4 stores and geography': {'owner': 'TODO', 'reviewer': 'TODO'},
        'Q5 signal or artifact':   {'owner': 'TODO', 'reviewer': 'TODO'},
    },
    'repo_url': 'TODO',
    'notebook_owner_this_week': 'TODO',         # one person commits the shared notebook
    'meeting_time': 'TODO',                     # a real recurring slot, not 'we will text'
    'channel': 'TODO',                          # where we talk, and how fast we answer
    'how_we_decide': 'TODO',                    # what we do when we disagree
    'if_someone_goes_quiet': 'TODO',
}

def find_todos(obj, path=()):
    if isinstance(obj, dict):
        return [t for k, v in obj.items() for t in find_todos(v, path + (k,))]
    if isinstance(obj, list):
        return [t for i, v in enumerate(obj, 1) for t in find_todos(v, path + (f'#{i}',))]
    return [' > '.join(path)] if str(obj).strip() in ('', 'TODO') else []

unfilled = find_todos(team)
if unfilled:
    print(f'{len(unfilled)} items still unfilled. First few:')
    for item in unfilled[:8]:
        print('  -', item)
else:
    print('nothing unfilled -- charter is complete')
````

````python
# Run this once the charter above has no unfilled items. It writes CHARTER.md next to this notebook.
def charter_markdown(t):
    lines = [f"# Team charter: {t['name']}", '', '## Members']
    lines += [f"- {m['name']} (GitHub: {m['github']})" for m in t['members']]
    lines += ['', '## Goals']
    lines += [f"- **{k.replace('_', ' ')}:** {v}" for k, v in t['goals'].items()]
    lines += ['', '## Roles', '', '| Role | Owner | Backup |', '|---|---|---|']
    lines += [f"| {r} | {v['owner']} | {v['backup']} |" for r, v in t['roles'].items()]
    lines += ['', '## Question owners', '', '| Question | Owner | Reviewer |', '|---|---|---|']
    lines += [f"| {q} | {v['owner']} | {v['reviewer']} |" for q, v in t['question_owners'].items()]
    lines += ['', '## How we work']
    for k in ('repo_url', 'notebook_owner_this_week', 'meeting_time',
              'channel', 'how_we_decide', 'if_someone_goes_quiet'):
        lines.append(f"- **{k.replace('_', ' ')}:** {t[k]}")
    return '\n'.join(lines) + '\n'

if unfilled:
    print('Finish the charter above first, then run this cell.')
else:
    with open('CHARTER.md', 'w') as f:
        f.write(charter_markdown(team))
    print('wrote CHARTER.md -- move it to the root of your repo and commit it')
````
